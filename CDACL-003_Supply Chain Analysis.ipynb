## EXPLORATORY DATA ANALYSIS (EDA)

```python
#Identify which products are most common.
plt.figure(figsize=(8,5))
_df['Product type'].value_counts().plot(kind='bar')
plt.title('Product Type Distribution')
plt.xlabel('Product Type')
plt.ylabel('Count')
plt.show()
```

```python
#Identify highest revenue generating product categories.
revenue = _df.groupby('Product type')['Revenue generated'].sum()

revenue.plot(kind='bar', figsize=(8,5))
plt.title('Revenue by Product Type')
plt.ylabel('Revenue')
plt.show()
```

```python
#Understand shipping expense distribution.
plt.figure(figsize=(8,5))
sns.histplot(_df['Shipping costs'], bins=20)
plt.title('Shipping Cost Distribution')
plt.show()
```

```python
#Identify best performing suppliers.
supplier = _df.groupby('Supplier name')['Revenue generated'].sum()

supplier.sort_values(ascending=False).head(10).plot(kind='bar')

plt.title('Top Suppliers by Revenue')
plt.show()
```

```python
#Detect low stock and overstock conditions.
plt.figure(figsize=(8,5))
sns.boxplot(x=_df['Stock levels'])
plt.title('Stock Level Distribution')
plt.show()
```

```python
#Understand production expenses.
cost = _df.groupby('Product type')['Manufacturing costs'].mean()

cost.plot(kind='bar')
plt.title('Average Manufacturing Cost')
plt.show()
```

```python
#Identify relation between cost and defects.
plt.figure(figsize=(8,5))
sns.scatterplot(x='Manufacturing costs', y='Defect rates', data=_df)
plt.title('Manufacturing Cost vs Defect Rate')
plt.show()
```

```python
#Understand transportation usage.
_df['Transportation modes'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Transportation Modes')
plt.ylabel('')
plt.show()
```

```python
#Identify delivery delay trends.
sns.lineplot(x=_df.index, y=_df['Shipping times'])
plt.title('Shipping Time Trend')
plt.show()
```

```python
#Analyze cost efficiency.
sns.scatterplot(x='Shipping costs', y='Revenue generated', data=_df)
plt.title('Revenue vs Shipping Cost')
plt.show()
```
